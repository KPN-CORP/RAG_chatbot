```python
import os
import google.generativeai as genai
```

```python

# 1. Configure API key
genai.configure(api_key=os.environ["GEMINI_API_KEY"])
```

```python

# 2. Pick a Gemini model
model = genai.GenerativeModel("gemini-2.5-flash")  # fast, cheaper
# model = genai.GenerativeModel("gemini-1.5-pro")   # deeper reasoning
```

```python

# 3. Run a prompt
response = model.generate_content("Apakah bidang usaha utama dari perusahaan KPN-corp")
print(response.text)
```
