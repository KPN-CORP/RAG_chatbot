```python
import os
```

```python
file = 'assets/CP-Bantuan-Pengobatan-2025.pdf'
```

```python
pdf_pages = extract_pdf_content(file)
```

```python
pdf_pages[2]
```

```python
filename = os.path.basename(file)
all_chunks = pages_to_chunks(pdf_pages, source=filename, language='id')
```

```python
all_chunks[0]
```

```python
for chunk in all_chunks:
    if "BANTUAN PENGOBATAN RAWAT JALAN" in chunk['chunk_text']:
        print(chunk['chunk_text'], '\n\n')
```

```python
for chunk in all_chunks:
    if chunk['page']==3:
        print(chunk['chunk_text'], '\n\n')
```

```python
for chunk in all_chunks:
    if chunk.get('table_id'):
        print(chunk)
        break
```
