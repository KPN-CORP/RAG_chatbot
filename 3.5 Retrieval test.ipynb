```python
from vectorstore.milvus_vectorstore import get_vectorstore
```

```python
# simple test snippet
vs = get_vectorstore()
retriever = vs.as_retriever(search_type="similarity", search_kwargs={"k": 5})
```

```python
query = 'Apa saja ketentuan cuti sakit?'
```

```python
docs = retriever.get_relevant_documents(query)
```

```python
docs
```

```python
print("Retrieved docs count:", len(docs))
for i, d in enumerate(docs, 1):
    print(f"--- doc {i} ---")
    print("page_content type:", type(d.page_content))
    print("page_content (preview):", d.page_content[:500])   # should be a string
    print("metadata:", d.metadata)
```
