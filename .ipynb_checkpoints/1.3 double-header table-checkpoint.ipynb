```python
def process_complex_table(table_data):
    """
    Processes a complex, multi-header table into a list of self-contained facts.
    """
    if len(table_data) < 3:
        # Not enough rows to be a complex table with data
        return []

    # Step 1: Reconstruct and Flatten the Headers
    header_row_1 = table_data[0]
    header_row_2 = table_data[1]
    
    # Forward-fill the top-level headers
    filled_header_1 = []
    last_header = ""
    for header in header_row_1:
        if header is not None:
            last_header = header
        filled_header_1.append(last_header)

    # Combine headers into a final, flat list
    flattened_headers = []
    for i, sub_header in enumerate(header_row_2):
        main_header = filled_header_1[i]
        if sub_header:
            # Clean up newlines and combine for a descriptive header
            clean_sub_header = sub_header.replace('\n', ' ')
            flattened_headers.append(f"{main_header} - {clean_sub_header}")
        else:
            flattened_headers.append(main_header)

    # Step 2: Unroll Each Cell into a Fact
    chunks = []
    data_rows = table_data[2:] # The rest of the list is data
    
    # Get the name of the primary identifier column (e.g., 'Gol.')
    row_identifier_name = flattened_headers[0]

    for row in data_rows:
        # The first item in the row is our primary key (e.g., '10', '9')
        row_identifier_value = row[0]
        if row_identifier_value is None:
            continue # Skip empty rows

        # Iterate through the rest of the cells in the row
        for i in range(1, len(row)):
            cell_value = row[i]
            if cell_value: # Only create a chunk if the cell has data
                header_context = flattened_headers[i]
                chunk_text = f"For {row_identifier_name} {row_identifier_value}, the value for '{header_context}' is {cell_value}."
                chunks.append(chunk_text)
                
    return chunks
```

```python
def process_complex_table(table_data):
    """
    Processes a complex, multi-header table into a list of dictionaries (records),
    where each dictionary represents a row.
    """
    if len(table_data) < 3:
        # Not enough rows for headers and data
        return []

    # --- Step 1: Reconstruct and Flatten the Headers (No changes here) ---
    header_row_1 = table_data[0]
    header_row_2 = table_data[1]
    
    # Forward-fill the top-level headers
    filled_header_1 = []
    last_header = ""
    for header in header_row_1:
        if header is not None:
            last_header = header
        filled_header_1.append(last_header)

    # Combine headers into a final, flat list
    flattened_headers = []
    for i, sub_header in enumerate(header_row_2):
        main_header = filled_header_1[i]
        if sub_header:
            clean_sub_header = sub_header.replace('\n', ' ')
            flattened_headers.append(f"{main_header} - {clean_sub_header}")
        else:
            flattened_headers.append(main_header)

    # --- Step 2: Build a List of Records (This part is new) ---
    records = []
    data_rows = table_data[2:] # The rest of the list is data

    for row in data_rows:
        # Skip any row that is empty or doesn't have a primary identifier
        if not row or row[0] is None:
            continue

        # Create a dictionary for the current row by zipping headers and row values
        # The dict() constructor combined with zip() is a concise way to do this.
        record = dict(zip(flattened_headers, row))
        records.append(record)
            
    return records

# --- Your example data ---
example_table = [
    ['Gol.', 'Dalam Negeri (dalam ribuan Rupiah)', None, None, None, None, 'Luar Negeri (dalam USD)', None],
    [None, 'Sumatera', 'Jawa & Bali', 'Kalimantan\n(selain Kaltim)', 'Kaltim &\nSulawesi', 'Papua', 'Asia1), Afrika\n(dalam USD)', 'Non Asia& Afrika,\nSingapore, Hongkong dan\nJepang (dalam USD)'],
    ['10', 'Sesuai tagihan, Hotel Bintang 5 (lima)', None, None, None, None, None, None],
    ['9', '1.200', '1.500', '1.200', '1.200', '1.500', '300', '400'],
    ['8', '1.000', '1.350', '1.000', '1.000', '1.350', '250', '350'],
    ['7', '900', '1200', '900', '900', '1.200', '200', '250'],
    ['6', '800', '900', '750', '800', '900', '80', '150'],
    ['5', '650', '750', '650', '700', '800', None, None],
    ['4', None, None, None, None, None, None, None],
    ['3', '450', '600', '400', '500', '600', None, None],
    ['2', '350', '500', '350', '400', '500', None, None]
]

# --- Run the processing and print the result ---
generated_records = process_table_to_records(example_table)

import json
print(json.dumps(generated_records, indent=2, ensure_ascii=False))
```

```python
def check_double_header(table_data):
    """
    Analyzes a table's structure and returns its type.
    """
    if len(table_data) < 2:
        return False # or "too_small"

    header_row_1 = table_data[0]
    header_row_2 = table_data[1]

    # Heuristic for the complex table:
    # Does the second row have non-None values under None values in the first?
    # This is a strong sign of a merged, hierarchical header.

    for i, item in enumerate(header_row_1):
        if item is None and header_row_2[i] is not None:
            return True
    
    # Heuristic for a simple table:
    # The first row is full and the second row looks like data.
    if all(isinstance(h, str) for h in header_row_1):
        return False
        
    return False # Default fallback
```
