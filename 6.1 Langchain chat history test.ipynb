```python
import requests
```

```python

# Create session
sid = requests.post("http://127.0.0.1:8000/session").json()["session_id"]
sid
```

```python

# Ask a question
resp = requests.post("http://127.0.0.1:8000/chat", json={
    "session_id": sid,
    "question": "Bagaimana jika saya ingin cuti karena sakit?"
}).json()

print(resp)
```
